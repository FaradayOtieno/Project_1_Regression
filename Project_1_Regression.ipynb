```python
# Prepare modelling dataset and split into training and testing sets

from sklearn.model_selection import train_test_split

# Remove identifiers and date column
model_df = df.drop(
    columns=["order_id", "customer_id", "order_date"]
).copy()

# Separate predictors (X) and target variable (y)
X = model_df.drop(columns=["revenue"])
y = model_df["revenue"]

# Split data: 80% training, 20% testing
X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.20,
    random_state=42
)

# Display dataset dimensions
print("Modelling dataset shape:", model_df.shape)
print("Training predictors:", X_train.shape)
print("Testing predictors:", X_test.shape)
print("Training target:", y_train.shape)
print("Testing target:", y_test.shape)
```

```python
# Linear Regression: preprocessing, training and prediction

from sklearn.compose import ColumnTransformer
from sklearn.preprocessing import OneHotEncoder
from sklearn.pipeline import Pipeline
from sklearn.linear_model import LinearRegression

# Identify numerical and categorical predictors
numeric_features = X_train.select_dtypes(include="number").columns
categorical_features = X_train.select_dtypes(include="object").columns

# Preprocess categorical variables
preprocessor = ColumnTransformer(
    transformers=[
        ("categorical", OneHotEncoder(drop="first", handle_unknown="ignore"),
         categorical_features)
    ],
    remainder="passthrough"
)

# Create Linear Regression pipeline
linear_model = Pipeline(steps=[
    ("preprocessor", preprocessor),
    ("regressor", LinearRegression())
])

# Train model
linear_model.fit(X_train, y_train)

# Generate predictions
y_pred_lr = linear_model.predict(X_test)

print("Linear Regression model trained successfully.")
print("Number of predictions:", len(y_pred_lr))
print("First 10 predictions:", y_pred_lr[:10])
```

```python
# Linear Regression: evaluation metrics and diagnostic plots

from sklearn.metrics import (
    mean_absolute_error,
    mean_squared_error,
    r2_score
)
import scipy.stats as stats

# Calculate evaluation metrics
lr_mae = mean_absolute_error(y_test, y_pred_lr)
lr_rmse = np.sqrt(mean_squared_error(y_test, y_pred_lr))
lr_r2 = r2_score(y_test, y_pred_lr)

print("LINEAR REGRESSION PERFORMANCE")
print("--------------------------------")
print(f"MAE:  {lr_mae:.4f}")
print(f"RMSE: {lr_rmse:.4f}")
print(f"R²:   {lr_r2:.4f}")

# Calculate residuals
residuals = y_test - y_pred_lr

# Create diagnostic plots
fig, axes = plt.subplots(2, 2, figsize=(13, 10))

# 1. Actual vs Predicted
axes[0, 0].scatter(y_test, y_pred_lr, alpha=0.5)
axes[0, 0].plot(
    [y_test.min(), y_test.max()],
    [y_test.min(), y_test.max()],
    linestyle="--",
    color="red"
)
axes[0, 0].set_title("Actual vs Predicted Revenue")
axes[0, 0].set_xlabel("Actual Revenue")
axes[0, 0].set_ylabel("Predicted Revenue")

# 2. Residuals vs Predicted
axes[0, 1].scatter(y_pred_lr, residuals, alpha=0.5)
axes[0, 1].axhline(0, linestyle="--", color="red")
axes[0, 1].set_title("Residuals vs Predicted Values")
axes[0, 1].set_xlabel("Predicted Revenue")
axes[0, 1].set_ylabel("Residuals")

# 3. Residual distribution
sns.histplot(residuals, kde=True, ax=axes[1, 0])
axes[1, 0].set_title("Distribution of Residuals")
axes[1, 0].set_xlabel("Residuals")

# 4. Residual Q-Q plot
stats.probplot(residuals, dist="norm", plot=axes[1, 1])
axes[1, 1].set_title("Q-Q Plot of Residuals")

plt.tight_layout()
plt.show()
```

```python
# Decision Tree Regression: preprocessing, training and prediction

from sklearn.tree import DecisionTreeRegressor, plot_tree
from sklearn.base import clone

# Create Decision Tree pipeline
tree_model = Pipeline(steps=[
    ("preprocessor", clone(preprocessor)),
    ("regressor", DecisionTreeRegressor(random_state=42))
])

# Train the model
tree_model.fit(X_train, y_train)

# Generate predictions
y_pred_tree = tree_model.predict(X_test)

print("Decision Tree model trained successfully.")
print("Number of predictions:", len(y_pred_tree))

# Display first 10 predictions
print("\nFirst 10 predictions:")
print(y_pred_tree[:10])

# Display tree complexity
trained_tree = tree_model.named_steps["regressor"]

print("\nDecision Tree structure:")
print("Number of leaves:", trained_tree.get_n_leaves())
print("Tree depth:", trained_tree.get_depth())
```

```python
# Random Forest Regression: training and prediction

from sklearn.ensemble import RandomForestRegressor

# Create Random Forest pipeline
rf_model = Pipeline(steps=[
    ("preprocessor", clone(preprocessor)),
    ("regressor", RandomForestRegressor(
        n_estimators=100,
        random_state=42,
        n_jobs=-1
    ))
])

# Train the model
rf_model.fit(X_train, y_train)

# Generate predictions
y_pred_rf = rf_model.predict(X_test)

print("Random Forest model trained successfully.")
print("Number of predictions:", len(y_pred_rf))

print("\nFirst 10 predictions:")
print(y_pred_rf[:10])

# Display Random Forest configuration
trained_rf = rf_model.named_steps["regressor"]

print("\nRandom Forest configuration:")
print("Number of trees:", trained_rf.n_estimators)
print("Maximum tree depth:", trained_rf.max_depth)
```

```python
# Compare Linear Regression, Decision Tree and Random Forest

from sklearn.metrics import mean_absolute_error, mean_squared_error, r2_score

# Store models and their test predictions
models = {
    "Linear Regression": (linear_model, y_pred_lr),
    "Decision Tree": (tree_model, y_pred_tree),
    "Random Forest": (rf_model, y_pred_rf)
}

results = []

for name, (model, predictions) in models.items():

    # Test-set performance
    test_mae = mean_absolute_error(y_test, predictions)
    test_rmse = np.sqrt(mean_squared_error(y_test, predictions))
    test_r2 = r2_score(y_test, predictions)

    # Training-set performance
    train_predictions = model.predict(X_train)
    train_r2 = r2_score(y_train, train_predictions)

    results.append({
        "Model": name,
        "MAE": test_mae,
        "RMSE": test_rmse,
        "Test R2": test_r2,
        "Train R2": train_r2,
        "R2 Gap": train_r2 - test_r2
    })

# Create comparison table
comparison_df = pd.DataFrame(results)

# Sort by lowest test RMSE
comparison_df = comparison_df.sort_values("RMSE").reset_index(drop=True)

print("MODEL PERFORMANCE COMPARISON")
display(comparison_df.round(4))

# Identify the best model by test RMSE
best_model_name = comparison_df.loc[0, "Model"]

print("\nBest model based on test RMSE:", best_model_name)
```

```python
# 10-Fold Cross-Validation for Random Forest

from sklearn.model_selection import KFold, cross_validate

# Define 10-fold cross-validation
cv = KFold(
    n_splits=10,
    shuffle=True,
    random_state=42
)

# Evaluate the Random Forest using multiple metrics
cv_results = cross_validate(
    rf_model,
    X_train,
    y_train,
    cv=cv,
    scoring={
        "MAE": "neg_mean_absolute_error",
        "RMSE": "neg_root_mean_squared_error",
        "R2": "r2"
    },
    return_train_score=True,
    n_jobs=1
)

# Organise fold-level results
cv_df = pd.DataFrame({
    "Fold": range(1, 11),
    "Validation MAE": -cv_results["test_MAE"],
    "Validation RMSE": -cv_results["test_RMSE"],
    "Validation R2": cv_results["test_R2"],
    "Training R2": cv_results["train_R2"]
})

print("10-FOLD CROSS-VALIDATION RESULTS")
display(cv_df.round(4))

# Summarise average performance and variability
print("\nCROSS-VALIDATION SUMMARY")
display(
    cv_df.drop(columns="Fold")
    .agg(["mean", "std"])
    .round(4)
)
```

```python
# Random Forest Hyperparameter Tuning

from sklearn.model_selection import GridSearchCV

# Define the parameter grid
param_grid = {
    "regressor__max_features": [0.5, 1.0],
    "regressor__min_samples_leaf": [1, 2],
    "regressor__max_depth": [None, 20]
}

# Set up grid search
rf_grid_search = GridSearchCV(
    estimator=rf_model,
    param_grid=param_grid,
    scoring="neg_root_mean_squared_error",
    cv=cv,
    refit=True,
    return_train_score=True,
    n_jobs=1,
    verbose=1
)

# Fit using training data only
rf_grid_search.fit(X_train, y_train)

# Extract tuning results
tuning_results = pd.DataFrame(rf_grid_search.cv_results_)

tuning_summary = tuning_results[[
    "param_regressor__max_features",
    "param_regressor__min_samples_leaf",
    "param_regressor__max_depth",
    "mean_test_score",
    "std_test_score",
    "mean_train_score"
]].copy()

# Convert negative RMSE scores into positive values
tuning_summary["mean_test_RMSE"] = -tuning_summary["mean_test_score"]
tuning_summary["std_test_RMSE"] = tuning_summary["std_test_score"]
tuning_summary["mean_train_RMSE"] = -tuning_summary["mean_train_score"]

tuning_summary = tuning_summary.drop(columns=[
    "mean_test_score",
    "std_test_score",
    "mean_train_score"
])

tuning_summary = tuning_summary.sort_values(
    "mean_test_RMSE"
).reset_index(drop=True)

print("TOP RANDOM FOREST HYPERPARAMETER COMBINATIONS")
display(tuning_summary.round(4))

print("\nBEST HYPERPARAMETERS:")
print(rf_grid_search.best_params_)

print("\nBEST CROSS-VALIDATION RMSE:")
print(round(-rf_grid_search.best_score_, 4))
```

```python
# Finalise the tuned Random Forest and evaluate on test data

from sklearn.metrics import mean_absolute_error, mean_squared_error, r2_score

# Retrieve the best fitted model
final_rf_model = rf_grid_search.best_estimator_

# Generate final predictions
y_pred_final_rf = final_rf_model.predict(X_test)

# Calculate final test metrics
final_mae = mean_absolute_error(y_test, y_pred_final_rf)
final_rmse = np.sqrt(mean_squared_error(y_test, y_pred_final_rf))
final_r2 = r2_score(y_test, y_pred_final_rf)

# Calculate training performance
y_train_pred_final = final_rf_model.predict(X_train)

final_train_mae = mean_absolute_error(y_train, y_train_pred_final)
final_train_rmse = np.sqrt(mean_squared_error(y_train, y_train_pred_final))
final_train_r2 = r2_score(y_train, y_train_pred_final)

# Final model performance table
final_metrics = pd.DataFrame({
    "Dataset": ["Training", "Testing"],
    "MAE": [final_train_mae, final_mae],
    "RMSE": [final_train_rmse, final_rmse],
    "R2": [final_train_r2, final_r2]
})

print("FINAL TUNED RANDOM FOREST PERFORMANCE")
display(final_metrics.round(4))

# Compare original models with the tuned Random Forest
final_comparison = pd.DataFrame([
    {
        "Model": "Linear Regression",
        "Test RMSE": lr_rmse,
        "Test R2": lr_r2
    },
    {
        "Model": "Decision Tree",
        "Test RMSE": mean_squared_error(y_test, y_pred_tree) ** 0.5,
        "Test R2": r2_score(y_test, y_pred_tree)
    },
    {
        "Model": "Random Forest (Original)",
        "Test RMSE": mean_squared_error(y_test, y_pred_rf) ** 0.5,
        "Test R2": r2_score(y_test, y_pred_rf)
    },
    {
        "Model": "Random Forest (Tuned)",
        "Test RMSE": final_rmse,
        "Test R2": final_r2
    }
]).sort_values("Test RMSE").reset_index(drop=True)

print("\nFINAL MODEL COMPARISON")
display(final_comparison.round(4))

print("\nSELECTED HYPERPARAMETERS:")
print(rf_grid_search.best_params_)
```
